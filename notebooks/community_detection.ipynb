```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# Define categories and file paths
categories = ["Government", "TV Shows", "Politician", "Company"]

file_paths = {
    "Government": "/Users/shrutisivakumar/Library/CloudStorage/OneDrive-Personal/College Stuff/Sem 1/Projects/Discrete Mathematics/facebook_clean_data/government_edges.csv",
    "TV Shows": "/Users/shrutisivakumar/Library/CloudStorage/OneDrive-Personal/College Stuff/Sem 1/Projects/Discrete Mathematics/facebook_clean_data/tvshow_edges.csv",
    "Politician": "/Users/shrutisivakumar/Library/CloudStorage/OneDrive-Personal/College Stuff/Sem 1/Projects/Discrete Mathematics/facebook_clean_data/politician_edges.csv",
    "Company": "/Users/shrutisivakumar/Library/CloudStorage/OneDrive-Personal/College Stuff/Sem 1/Projects/Discrete Mathematics/facebook_clean_data/company_edges.csv"
}
```

```python
# Function to load data from CSV and create a graph
def load_data(file_path):
    df = pd.read_csv(file_path)
    G = nx.from_pandas_edgelist(df, "node_1", "node_2")
    return G
```

```python
# Function for community detection and visualization
def detect_communities(G, category):
    communities = list(nx.algorithms.community.greedy_modularity_communities(G))
    num_communities = len(communities)
    print(f"Number of communities in {category} category: {num_communities}")

    # Set a threshold for visualizing communities based on the number of nodes
    if len(G.nodes) < 10000:
        # Visualize each community separately
        for i, community in enumerate(communities):
            subgraph = G.subgraph(community)
            if len(subgraph.nodes) < 1000:
                pos = nx.spring_layout(subgraph)
                colors = [i] * len(subgraph.nodes)
                nx.draw(subgraph, pos, node_color=colors, with_labels=True)
                plt.title(f"Community {i+1} in {category} Category")
                plt.show()
```

```python
# Iterate through categories and perform community detection
for category in categories:
    print(f"\n--- {category} ---")
    G = load_data(file_paths[category])
    detect_communities(G, category)
```
